```python
#!/usr/bin/env python3
from pathlib import Path

def rewrite_paths(infile, outfile,
                  new_base="/home/ardhendu/log-logistics/FEATURES_DIRECTORY1",
                  old_base="/media/ardhendu/a5dea57f-ec8a-43f0-a2ba-f253dff96634/ardhendu/sgcmde/FEATURES_DIRECTORY1",
                  add_header_line=True,
                  drop_csv_header=True):
    infile = Path(infile)
    outfile = Path(outfile)

    with infile.open("r", encoding="utf-8") as f:
        lines = f.read().splitlines()

    # Optionally drop the CSV header row
    if drop_csv_header and lines and "," in lines[0]:
        # first row looks like 'case_id,wsi_id,label,feature_h5_path'
        lines = lines[1:]

    # Replace old base path with new base path in every remaining line
    rewritten = [ln.replace(old_base, new_base) for ln in lines if ln.strip()]

    with outfile.open("w", encoding="utf-8") as f:
        if add_header_line:
            f.write(new_base + "\n\n")  # base path line + blank line
        for ln in rewritten:
            f.write(ln.rstrip() + "\n")

if __name__ == "__main__":
    # Example usage:
    # rewrite_paths("in.txt", "out.txt")
    # Adjust paths/bases below if needed.
    rewrite_paths(
        infile="wsi_annos_vit-s-dino-p16.txt",
        outfile="out.txt",
        new_base="/home/ardhendu/log-logistics/FEATURES_DIRECTORY1",
        old_base="/media/ardhendu/a5dea57f-ec8a-43f0-a2ba-f253dff96634/ardhendu/sgcmde/FEATURES_DIRECTORY1",
        add_header_line=True,   # writes the base-path line at top
        drop_csv_header=True    # removes the CSV header row
    )
```
