```python
in_path  = "wsi_annos_vit-s-dino-p16.txt"   # your input comma-separated text file
out_path = "wsi_annos_vit-s-dino-p17.txt"

old = "/media/ardhendu/a5dea57f-ec8a-43f0-a2ba-f253dff96634/ardhendu/log-logistics/FEATURES_DIRECTORYBRCA/h5_files"
new = "/home/ardhendu/log-logistics/FEATURES_DIRECTORYBRCA/h5_files"

with open(in_path, "r", encoding="utf-8") as f:
    lines = [line.rstrip("\n") for line in f]

header = lines[0].split(",")
try:
    idx = header.index("feature_h5_path")
except ValueError:
    raise RuntimeError("Column 'feature_h5_path' not found in header.")

out_lines = [",".join(header)]
for line in lines[1:]:
    parts = line.split(",")
    if len(parts) > idx:
        parts[idx] = parts[idx].replace(old, new)
    out_lines.append(",".join(parts))

with open(out_path, "w", encoding="utf-8") as f:
    f.write("\n".join(out_lines) + "\n")

print(f"Saved: {out_path}")
```
